```python
from unsloth import FastModel
import torch
from huggingface_hub import snapshot_download

max_seq_length = 2048 # Choose any for long context!

fourbit_models = [
    # 4bit dynamic quants for superior accuracy and low memory use
    "unsloth/gemma-3-4b-it-unsloth-bnb-4bit",
    "unsloth/gemma-3-12b-it-unsloth-bnb-4bit",
    "unsloth/gemma-3-27b-it-unsloth-bnb-4bit",
    # Qwen3 new models
    "unsloth/Qwen3-4B-unsloth-bnb-4bit",
    "unsloth/Qwen3-8B-unsloth-bnb-4bit",
    # Other very popular models!
    "unsloth/Llama-3.1-8B",
    "unsloth/Llama-3.2-3B",
    "unsloth/Llama-3.3-70B",
    "unsloth/mistral-7b-instruct-v0.3",
    "unsloth/Phi-4",
] # More models at https://huggingface.co/unsloth

# Download model and code
snapshot_download("SparkAudio/Spark-TTS-0.5B", local_dir = "Spark-TTS-0.5B")

model, tokenizer = FastModel.from_pretrained(
    model_name = "MohamedGomaa30/spark-tts-normazlied-masri-mega",
    max_seq_length = max_seq_length,
    dtype = torch.float32, # Spark seems to only work on float32 for now
    full_finetuning = True, # We support full finetuning now!
    load_in_4bit = False,
    #token = "hf_...", # use one if using gated models like meta-llama/Llama-2-7b-hf
)
```

```python
from datasets import load_dataset 
dataset = load_dataset("MohamedGomaa30/MasriAudio-Mega-sparkTTS", split="train")
```

```python
from trl import SFTConfig, SFTTrainer
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    dataset_text_field = "text",
    max_seq_length = max_seq_length,
    packing = False, 

    args = SFTConfig(
        per_device_train_batch_size = 8,
        gradient_accumulation_steps = 2,
        warmup_steps = 500,
        num_train_epochs = 15,
        learning_rate = 2e-5,  
        fp16 = False,
        bf16 = True, 
        logging_steps = 100,
        
            # --- Gradient Norm Applied ---
        max_grad_norm = 1.0,  
    
        # --- Checkpoint & Hub Settings ---
        push_to_hub = True,
        hub_model_id = "MohamedGomaa30/spark-tts-normazlied-masri-mega", # Ensure this is set
        hub_strategy = "checkpoint",               # Push to hub every time we save
        save_strategy = "steps",
        save_steps = 1000,                         # Save/Push every 500 steps
        save_total_limit = 2,                      # Keep only the 2 latest checkpoints (saves space)
        load_best_model_at_end = False,            # Set True if using evaluation_strategy
        # ---------------------------------
        optim = "adamw_8bit", #"adamw_8bit"
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "spark-tts-normazlied-masri-mega",
        report_to = "wandb",
    ),
)
```

```python
# @title Show current memory stats
gpu_stats = torch.cuda.get_device_properties(0)
start_gpu_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
max_memory = round(gpu_stats.total_memory / 1024 / 1024 / 1024, 3)
print(f"GPU = {gpu_stats.name}. Max memory = {max_memory} GB.")
print(f"{start_gpu_memory} GB of memory reserved.")
```
